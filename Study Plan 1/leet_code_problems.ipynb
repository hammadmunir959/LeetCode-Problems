<b> 1.  Merging Strings Alternately </b>

```python
class Solution:

    def mergeAlternately(self, word1: str, word2: str) -> str:
        result = []

        i,j = 0,0
        while i < len(word1) and j < len(word2):
            result.append(word1[i])
            result.append(word2[j])
            j += 1
            i += 1
        
        if j < len(word2):
            result.append(word2[j:])

        elif i < len(word1):
            result.append(word1[i:])
        else:
            pass
        
        return ''.join(result)

word1 = 'Hammad '
word2 = 'Munir'

problem = Solution()

print(problem.mergeAlternately(word1, word2))

        
```

<b>Find the Difference </b>

You are given two strings s and t.

String t is generated by random shuffling string s and then add one more letter at a random position.

Return the letter that was added to t.

```python

class Solution:
    def find_the_difference(self, s : str , t: str) -> str:
        result = 0
        for m in s:
            result = result ^ ord(m)
        
        for n in t:
            result = result ^ ord(n)

        return chr(result)

            
string1 = str(input("Enter a string1 : ")).lower()
string2 = str(input("Enter a string2 : ")).lower()

print(f"string 1 = {string1},  string 2 = {string2}")

problem = Solution()

print(f" difference =  {problem.find_the_difference(string1, string2)}")
```
